# 02 Fetch and Join\n\nFetch data from two sources and join in pandas.

```python
from policy_data_ai.tools.api import PolicyDataRuntime\n\nruntime = PolicyDataRuntime(registry_dir="../sources_registry")
```

```python
wb = runtime.tool_fetch_data(\n    "worldbank",\n    "SL.UEM.TOTL.ZS",\n    filters={"country": ["DE", "FR"]},\n    start="2010",\n    end="2024",\n)\nwb_df = wb.data\nwb_df.head()
```

```python
owid = runtime.tool_fetch_data(\n    "owid",\n    "population",\n    filters={"Entity": ["Germany", "France"]},\n    start="2010",\n    end="2024",\n)\nowid_df = owid.data\nowid_df.head()
```

```python
wb_small = wb_df[["country_name", "year", "value"]].rename(columns={"value": "unemployment_rate"})\nowid_small = owid_df[["Entity", "Year", "population"]].rename(columns={"Entity": "country_name", "Year": "year"})\njoined = wb_small.merge(owid_small, on=["country_name", "year"], how="left")\njoined.head()
```
